## Implement the Rules

```r
classify_Eligibility <- function (Programming_Grade, Math_Grade, Attendance, Prerequisite){
  if (Programming_Grade >= 80 && Math_Grade >= 75 && Attendance >= 80 && Prerequisite == "Yes"){
  print("Eligible")
} else if (Programming_Grade >= 75 && Math_Grade >= 70 && Attendance >= 75) {
  print("Conditionally Eligible")
} else {
  print("Not Eligible")
}
}
```

## Apply the Functions to All Students

```r
students$Eligbility <- mapply(
  classify_Eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite

)

students
```

## Create Recommendation Rules

```r
student_Recommendations <- function(classify_Eligibility)
{
  if (classify_Eligibility == "Eligible") {
  return ("Student May Enroll")
} else if (classify_Eligibility == "Conditionally Eligible"){
  return ("Consult Academic Tutor")
} else if (classify_Eligibility == "Not Eligible"){
  return ("Complete Requirements Before Enrollment")
}
}

students$Recommendation <- sapply(
  students$Eligbility,
  student_Recommendations
)

students
```
