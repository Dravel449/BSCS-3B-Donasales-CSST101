```r
give_recommendation <- function(classify_Risk)
{
  if (classify_Risk == "HIGH RISK") {
  return ("Immediate academic intervention")
} else if (classify_Risk == "Moderate Risk"){
  return ("Academic monitoring")
} else if (classify_Risk == "Low Risk"){
  return ("Continue regular monitoring")
}
}

give_recommendation("HIGH RISK")
```
